```python
import tensorflow as tf
import numpy as np
```

```python
celcius = np.array([-20, 100, 50,45,-42,30,13])
kelvin = celcius + 273.15

print(kelvin)
```

```python
capa_oculta_1 = tf.keras.layers.Dense(units=3, input_shape=[1])
capa_oculta_2 = tf.keras.layers.Dense(units=3)
Salida = tf.keras.layers.Dense(units=1)
model =tf.keras.Sequential([capa_oculta_1, capa_oculta_2, Salida])

model.compile(
    optimizer=tf.keras.optimizers.Adam(0.1),
    loss='mean_squared_error',
)
model.summary()
```

```python
historicla = model.fit(celcius, kelvin, epochs=500)
print("Modelo Entrenado")
```

```python
test_values =np.array([-20,0,50,120,67], dtype=float)

prdicciones = model.predict(test_values)

print("Celcius | Kelvin Real | Prediccion NN")
for c, p in zip(test_values, prdicciones):
    print(f"{c:7.2f} | {c+273.15:12.2f} | {float(p):.2f}")
```
