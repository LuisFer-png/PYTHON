```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/Pokemon.csv')
```

```python
stats =df[['Name','Type 1','Type 2','HP']]
stats.head(20)
```

```python
df['num_tipos'] = df['Type 2'].apply(lambda x: 1 if pd.isna(x) else 2)

conteo = df['num_tipos'].value_counts()

print(conteo)

labels = ['Un solo tipo', 'Dos tipos']

colors = ['#0026FF', '#FF0000']

plt.figure(figsize=(6,6))
plt.pie(conteo, labels=labels,autopct='%1.1f%%', colors=colors, startangle=180, textprops={'fontsize': 11}, wedgeprops={"edgecolor": 'white'})

plt.title('proporción de Pokemon de un tipo y dos tipos')
plt.show()
```

```python
pokemon_name = 'Arcanine'

pokemon = df[df['Name'] == pokemon_name].iloc[0]

stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
values =[pokemon[stat] for stat in stats]
values += values[:1]
angles = np.linspace(0, 2*np.pi, len(stats), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(10, 10))
plt.polar(angles, values, color='gold', linewidth=2)
plt.fill(angles, values, color='gold', alpha=0.2)

plt.xticks(angles[:-1], stats, color='gray', size=10)
plt.title(f'Estadísticas de {pokemon_name}', size=20)
plt.yticks(color='gray', size=10)
plt.ylim(0, max(values))
```
